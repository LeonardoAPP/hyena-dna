```python
from src.dataloaders.datasets.a_thaliana_dataset import FastaInterval

fa_file_path = 'data/A.Thaliana/TAIR10_mod.fna'

fi = FastaInterval(fasta_file=fa_file_path, pad_only=True)
seq = fi(chr_name="Chr1", start=10, end=20, max_length=40)  # end-start < max_length

print(seq)
print("Longitud:", len(seq))
print("Padding izquierdo:", seq.startswith("." * ((40 - (20-10)) // 2)))
print("Padding derecho:", seq.endswith("." * ((40 - (20-10)) - ((40 - (20-10)) // 2))))
```
